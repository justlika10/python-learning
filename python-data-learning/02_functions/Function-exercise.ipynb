```python
data = {
    "Netherlands": {
        "population": 17800,
        "gdp": 1000
    },
    "Georgia": {
        "population": 3710,
        "gdp": 80
    },
    "China": {
        "population": 1409000,
        "gdp": 17700
    }
}
```

```python
def get_country_info(name, di):
    return name, di[name]
        
print(get_country_info('Netherlands', data))
print(get_country_info('Georgia', data))
print(get_country_info('China', data))
            
```

```python
def find_largest_gdp(slov):
    maximus = 0
    for cou, infor in slov.items():
        if infor['gdp'] > maximus:
            maximus = infor['gdp']
            maximus_cou = cou
    return maximus_cou
    
print(find_largest_gdp(data))   
```

```python
data = {
    "Netherlands": {
        "2021": 900,
        "2022": 950,
        "2023": 1000
    },
    "Georgia": {
        "2021": 65,
        "2022": 75,
        "2023": 80
    },
    "China": {
        "2021": 17800,
        "2022": 17900,
        "2023": 17700
    }
}
```

```python
def calculate_growth(slovar):
    slovar_new = {}
    for country, statistics in slovar.items():
        diff = statistics['2023'] - statistics['2021']
        slovar_new[country] = diff
    return slovar_new
print(calculate_growth(data))    
```

```python
values = [100, 200, 300, 0, 450, -20, 500]
```

```python
def check_data(spisok):
    new_spisok = {}
    count_pos = 0
    count_im = 0
    count_zero = 0
    for i in spisok:
        if i > 0:
            count_pos += 1
        elif i == 0:
            count_zero += 1
        elif i < 0:
            count_im += 1
    new_spisok['Positive:'] = count_pos
    new_spisok['Zeroes'] = count_zero
    new_spisok['Impositive'] = count_im
    return new_spisok
    
print(check_data(values))
```
