```python
revenue = {
    "January": 4500,
    "February": 5200,
    "March": 3800,
    "April": 6100,
    "May": 7200,
    "June": 4900,
    "July": 8500
}
```

```python
all_revenue = sum(revenue.values())
all_revenue
```

```python
average_revenue = all_revenue // len(revenue.values())
average_revenue
```

```python
max(revenue, key=revenue.get)
```

```python
min(revenue, key = revenue.get)
```

```python
high_revenue = {}
for keys, values in revenue.items():
    if values > average_revenue:
        high_revenue[keys] = values
        
high_revenue
```

```python
count = 0
for first, second in revenue.items():
    if second > 5000:
        count += 1
count
```

```python
user_ages = {
    "user_1": 17,
    "user_2": 25,
    "user_3": -1,
    "user_4": 140,
    "user_5": 32,
    "user_6": 0,
    "user_7": 19,
    "user_8": -5,
    "user_9": 45,
    "user_10": 200
}
```

```python
valid_users = {}
invalid_users = {}
for name, age in user_ages.items():
    if age > 0 and age < 120:
        valid_users[name] = age
    else:
        invalid_users[name] = age
valid_users, invalid_users
```

```python
average_age = sum(valid_users.values()) // len(valid_users.values())
average_age
```

```python
max(valid_users, key=valid_users.get)
```

```python
min(valid_users, key=valid_users.get)
```

```python
perc = len(invalid_users.values()) / len(user_ages.values()) * 100
perc
```
