```python
data = {
    "Netherlands": {
        "2021": {"population": 17500, "gdp": 900},
        "2022": {"population": 17600, "gdp": 950},
        "2023": {"population": 17800, "gdp": 1000}
    },
    "Georgia": {
        "2021": {"population": 3700, "gdp": 65},
        "2022": {"population": 3700, "gdp": 75},
        "2023": {"population": 3710, "gdp": 80}
    },
    "China": {
        "2021": {"population": 1412000, "gdp": 17800},
        "2022": {"population": 1412000, "gdp": 17900},
        "2023": {"population": 1409000, "gdp": 17700}
    }
}
```

```python
for k, v in data.items():
    for year, g in v.items():
        if year == "2023":
            print(k, g['gdp'])
```

```python
max_2023 = 0
for k, v in data.items():
    for year, g in v.items():
        if year == "2023":
            if g['gdp'] > max_2023:
                max_2023 = g['gdp']
                max_cou = k
max_cou
```

```python
for k, v in data.items():
    for year, g in v.items():
        if year == '2021':
            second = g['gdp']
        elif year == '2023':
            first = g['gdp']
            print(k, first - second)
```

```python
max_diff = 0
for k, v in data.items():
    for year, g in v.items():
        if year == '2021':
            second = g['gdp']
        elif year == '2023':
            first = g['gdp']
            lolo = first - second
            if lolo > max_diff:
                max_diff = lolo
                max_diff_cou = k
                print(max_diff_cou)
```

```python
results = {max_diff_cou : max_diff}
results
```
