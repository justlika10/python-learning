```python
with open(r'C:\Users\V.B.Zaalishvili\Desktop\countries.json', encoding='utf-8') as file:    
    data = json.load(file)
    difference = {}
    for country, general in data.items():
        for year, info in general.items():
            if country == 'Netherlands':
                first = general['2023']['gdp']
                second = general['2021']['gdp']
                diff = first - second
                difference[country] = diff
            elif country == 'Georgia':
                first_g = general['2023']['gdp']
                second_g = general['2021']['gdp']
                diff_g = first_g - second_g
                difference[country] = diff_g
            elif country == 'China':
                first_c = general['2023']['gdp']
                second_c = general['2021']['gdp']
                diff_c = first_c - second_c
                difference[country] = diff_c
    print(difference)            
```

```python
with open(r'C:\Users\V.B.Zaalishvili\Desktop\countries.json', encoding='utf-8') as file:    
    data = json.load(file)
    def get_average_gdp(data_json, country):
        average = (data_json[country]['2021']['gdp'] + data_json[country]['2022']['gdp'] + data_json[country]['2023']['gdp']) / 3
        return average
    print(get_average_gdp(data, 'Georgia'))
```
