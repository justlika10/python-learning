```python
with open(r'C:\Users\V.B.Zaalishvili\Desktop\countries.csv', encoding='utf-8') as file:
    reader = csv.reader(file, delimiter=';')
    next(reader)
    average_gdp = {}
    totali_1 = 0
    totali_2 = 0
    totali_3 = 0
    for middle in reader:
        if middle[0] == 'Netherlands':
            middle[3] = int(middle[3])
            totali_1 += middle[3] 
            aver_1 = totali_1 / 3
            average_gdp['Netherlands'] = aver_1
        elif middle[0] == 'Georgia':         
            middle[3]= int(middle[3])
            totali_2 += middle[3]
            aver_2 = totali_2 / 3
            average_gdp['Georgia'] = aver_2
        elif middle[0] == 'China':
            middle[3] = int(middle[3])
            totali_3 += middle[3]
            aver_3 = totali_3 / 3
            average_gdp['China'] = aver_3
    print(average_gdp) 
```
